# PNG Tile Slicer

配置源 PNG、切割尺寸和输出目录，然后运行下面的单元格。纯透明 tile 会默认跳过，输出文件按 `0.png`, `1.png`, ... 连续编号。

```python
from pathlib import Path
import sys

# Notebook lives in python_town/notebooks, so add python_town to sys.path.
PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent
elif (PROJECT_ROOT / "python_town").exists():
    PROJECT_ROOT = PROJECT_ROOT / "python_town"

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from tools.slice_png_tiles import slice_png_tiles

# ===== 配置区 =====
input_png = r"F:\Documents\GitHub\AI_Town_Demo\art\Interiors\4_User_Interface_Elements\UI_32x32.png"
output_dir = r"F:\Documents\GitHub\AI_Town_Demo\art\character\emotion"
tile_width = 32
tile_height = 64
start_index = 0
skip_transparent = True
include_partial = False
clean_output_pngs = False
write_metadata = True

results = slice_png_tiles(
    input_path=input_png,
    output_dir=output_dir,
    tile_width=tile_width,
    tile_height=tile_height,
    start_index=start_index,
    skip_transparent=skip_transparent,
    include_partial=include_partial,
    clean_output_pngs=clean_output_pngs,
    write_metadata=write_metadata,
)

print(f"Saved {len(results)} tiles to {Path(output_dir).resolve()}")
results[:10]
```
